```python
import pandas as pd
```

```python
df = pd.read_csv("Airbnb_Open_Data.csv") 
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
cols_to_keep = [ 'NAME', 'host id', 'host_identity_verified', 'host name',
       'neighbourhood group', 'neighbourhood', 'lat', 'long', 'country',
       'country code', 'instant_bookable', 'cancellation_policy', 'room type',
       'Construction year', 'price', 'service fee', 'minimum nights',
       'number of reviews', 'last review'] 
cols_to_drop = ['license' , 'id','reviews per month','review rate number', 'calculated host listings count',
       'availability 365', 'house_rules']
```

```python
len(cols_to_keep)
```

```python
len(cols_to_drop)
```

### First Method is to filterring data

```python
df[cols_to_keep]
```

### Seconed Method Dropping Data

```python
df = df.drop(columns = df[cols_to_drop])
```

```python
df
```

```python
df.columns
```

```python
df.rename(columns = {
    'NAME' : 'name' } , inplace  = True)
```

```python
df.columns
```

```python
for i in df.columns:
    print(i.upper())
```

```python
df.duplicated()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated()
```

```python
df.isna().sum()
```

```python
df.drop(columns = [ 'last review'] , inplace = True)
```

```python
df.dropna(inplace = True)
```

```python
df.isna().sum()
```

```python
df['host_identity_verified'] = df['host_identity_verified'].str.upper()
```

```python
df.head()
```

```python
df['instant_bookable'] = df['instant_bookable'].apply(lambda x: 1 if x == True else 0)
```

```python
df['instant_bookable']
```

```python
df.head()
```

```python
df['price'] = df['price'].str.replace('$' , '')
```

```python
df['price'] = df['price'].str.replace(',' , '')
```

```python
df['price']= df['price'].str.replace(' ', '')
```
