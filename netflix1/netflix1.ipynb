```python
df.head(1)
```

```python
df.drop(['rating'], axis=1, inplace=True) 
```

```python
df.columns
```

```python
new_cols =  [] 
for i in df.columns:
    new_cols.append(i.replace(' ', '_').capitalize())

new_cols
df.columns = new_cols
df.columns
```

```python
df.duplicated().sum()
```

```python
df.isna().sum()
```

```python
df['Show_id'].nunique()
```

```python
df['Show_id'] = df['Show_id'].apply(lambda x: x.split('s')[1])
```

```python
df.head()
```

```python
df['Show_id'] = df['Show_id'].astype(int)
```

```python
df['Date_added'] = pd.to_datetime(df['Date_added'])
```

```python
df['Country'].value_counts().reset_index().head()
```
