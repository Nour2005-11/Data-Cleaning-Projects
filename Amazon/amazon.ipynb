```python
import pandas as pd
```

```python
df = pd.read_csv('amazon_product.csv')
```

```python
df.head()
```

```python
df.drop(['Unnamed: 0'], axis=1, inplace=True)
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.columns
```

```python
cols_drop = ['product_original_price' , 'product_availability' , 'unit_price' , 'unit_count']
df.drop(cols_drop, axis=1, inplace=True)
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
bool_cols = df.select_dtypes(include='bool').columns

df[bool_cols] = df[bool_cols].map(lambda x: 1 if x else 0)
```

```python
df.head()
```

```python
# Remove unnecessary columns
df.drop(columns=['product_photo', 'product_url'], inplace=True)

# Convert price columns to numeric
df['product_price'] = df['product_price'].str.replace('$', '').astype(float)

df['product_minimum_offer_price'] = (
    df['product_minimum_offer_price']
    .str.replace('$', '')
    .astype(float)
)

# Check currency
df['currency'].value_counts()
```

```python
df.info()
```

```python
df.nunique()
```

```python
df.columns = df.columns.str.capitalize()
df.columns
```

```python
df.to_csv('amazon_product_cleaned.csv')
df.to_csv('amazon_product_cleaned.csv', index=False)
```
